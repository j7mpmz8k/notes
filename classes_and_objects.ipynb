```python
#private variables (new method)
#getter methods can be called without () after the variable name.
#introduces deletable attributes

#NOTE new method with decorators negates mangling. Still reccomended for conventional reasons as clear signal to devs
class Rectangle:
    def __init__(self, pos:tuple[int, int]=(0,0), width:int=0, height:int=0):
        self.__useless_var = "This is a useless variable to be deleted."
        self.__x = pos[0]
        self.__y = pos[1]
        self.__width = width
        self.__height = height

    @center.setter
    def center(self, pos:tuple[int, int]):
        """called with obj.center = (x,y) instead of obj.center(x,y)"""
        self.__x = pos[0] - (self.__width / 2)
        self.__y = pos[1] - (self.__height / 2)

    @property
    def center(self):
        """called with obj.center instead of obj.center()"""
        return (self.__x + (self.__width / 2), self.__y + (self.__height / 2))

    @property
    def pos(self):
        """called with obj.pos instead of obj.pos()"""
        return (self.__x, self.__y)

    @pos.setter
    def pos(self, pos:tuple[int, int]):
        """called with obj.pos = (x,y) instead of obj.pos(x,y)"""
        self.__x = pos[0]
        self.__y = pos[1]


    @useless_var.deleter
    def useless_var(self):
        """
        called using 'del obj.useless_var'
        """
        print("Deleting the useless variable.")  # Optional custom logic
        del self.__useless_var
```

```python
#private variables (old method)

#NOTE __ triggers mangling unlike single _ (also works to mangle methods themselves)
#mangling renames the variable to _ClassName__variable for access outside of class ie. obj._ClassName__variable = new_value
class Rectangle:
    def __init__(self, pos:tuple[int, int]=(0,0), width:int=0, height:int=0):
        self.__x = pos[0]
        self.__y = pos[1]
        self.__width = width
        self.__height = height

    def set_center(self, pos:tuple[int, int]):
        self.__x = pos[0] - (self.__width / 2)
        self.__y = pos[1] - (self.__height / 2)

    def get_center(self):
        return (self.__x + (self.__width / 2), self.__y + (self.__height / 2))

    def get_pos(self):
        return (self.__x, self.__y)

    def set_pos(self, pos:tuple[int, int]):
        self.__x = pos[0]
        self.__y = pos[1]
```

```python
class Square:
    def __init__(self, side):
        self.width = side
    def area(self):
        return self.width ** 2
    def perimeter(self):
        return 4 * self.width
    def radius(self):
        return self.width / 2

class Rectangle(Square):
    def __init__(self, height, width):#since Rectangle added and extra parameter, must recreate __init__
        super().__init__(width)#without this line, width would have to be recalculated
        self.height = height
    def area(self):
        return self.height * self.width
    def perimeter(self):
        return 2 * (self.height + self.width)

class Triangle(Rectangle):
    """45:45:90 triangle"""
    #__init__ not needed since no extra parameters from Rectangle
    def area(self):
        return super().area() / 2
    def perimeter(self):
        return super().perimeter() / 2

triangle1 = Triangle(40,4)
print(triangle1.radius())#inherited from Square
print(triangle1.area())#overwritten and modified from from Rectangle
print(triangle1.perimeter())#overwritten and modified from from Rectangle
```

```python
class rectangle:
    #creating a method ie. calc_area
    def calc_area(self):
        return self.x * self.y

rectangle_a = rectangle()
rectangle_a.x = 100
rectangle_a.y = 200

rectangle_b = rectangle()
rectangle_b.x = 400
rectangle_b.y = 300

print(rectangle_a.calc_area())
```

```python
class rectangle:
    #__init__ method is special because it runs automatically when the object is created
    #benefits of this may include robust code enforcing uniform object creation
    def __init__(self, x:int, y:int):
        self.x = x
        self.y = y
        self.area = x * y if self.y and self.x else None#optional, "if self.y" returns True if not empty, False if empty

rectangle_a = rectangle(x=5, y=9)
rectangle_b = rectangle(x=35, y=99)

print(rectangle_a.area)
```
