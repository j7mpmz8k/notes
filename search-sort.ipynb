```python
#linear search
def main():
    # List does not have to be in order
    list_1 = [4, 6, 1, 43, 1, 76, 85, 30, 8, 29]
    key = 85
    result = linear_search(list_1, key)
    if result == -1:
        print("Did not find", key)
    else:
        print("Found,", key, "at position", result)


def linear_search(input_list, key):
    for i in range(len(input_list)):
        if key == input_list[i]:
            return i
    return -1
main()
```

```python
#binary search
def main():
    # List needs to be in order
    list_1 = [1, 4, 6, 8, 29, 30, 43, 76, 85]
    key = 43
    result = binary_search(list_1, key)
    if result == -1:
        print("Did not find", key)
    else:
        print("Found,", key, "at position", result)


def binary_search(input_list, key):
    print(input_list)
    low = 0
    high = len(input_list) - 1
    while high >= low:# = is needed for the high to eventually be checked
        print(low, high)
        mid = (high + low) // 2#recalculates middle index based on refactored low and high range
        if key == input_list[mid]:#checks middle index for match
            return mid
        elif key < input_list[mid]:
            high = mid - 1#refocus search to left half of checked index at middle
        else:
            low = mid + 1#refocus search to right half of checked index at middle

    return -1


main()
```

```python
#selection sort
import random

def main():
    lst = []

    for i in range(30):
        lst.append(random.randint(100, 999))

    print(lst)
    selection_sort(lst)
    print(lst)


def selection_sort(input_list):
    """
    i is next index to sort from left to right
    j is next unsorted index to the right of i to find a minimimum value less than than i moving left to right
    if j is less than i, curr_min_index is temporarily updated to j
    at the end of the inner loop, curr_min_index and i swap places
    """
    for i in range(len(input_list) - 1):# sorts each index from left to right one by one
        curr_min_index = i#initialize next minimum value to index being sorted

        for j in range(i + 1, len(input_list)):# Inner loop: scan unsorted portion
            if input_list[curr_min_index] > input_list[j]:#compare if checked index is new minimum value
                curr_min_index = j#update minimum index to index being checked

        if curr_min_index != i:#swap index of new minimum value with index being sorted asuming a new minimum value was found
            input_list[i], input_list[curr_min_index] = input_list[curr_min_index], input_list[i]


main()
```

```python
#insertion sort
import random

def main():
    lst = []

    for i in range(30):
        lst.append(random.randint(100, 999))

    print(lst)
    insertion_sort(lst)
    print(lst)


def insertion_sort(input_list):
    """
    i is the index of the element being sorted from left to right
    i is temporarily stored in curr_element since j gets copid to the right
    j is the previously sorted index to the left of i moving from right to left
    when j is less than curr_element, j is copied to the right of itself
    if curr_element is less than j, curr_element is inserted to the right of j(happens only once since j keeps shifting to the right until index curr_element to be placed is found)
    """
    for i in range(1, len(input_list)):
        curr_element = input_list[i]#stores next current element to sort from left to right...needed since j(previous index) gets shifted to the right
        j = i - 1#finds next previous index to compare with current element to sort...continues right to left until j is less than curr_element

        while j >= 0 and input_list[j] > curr_element:
            input_list[j + 1] = input_list[j]#copies previously sorted element (ie. j) to the right only if greater than curr_element
            j -= 1#updates j to next previously sorted element to then compare to curr_element

        input_list[j + 1] = curr_element#inserts curr_element to right of j


main()
```

```python
#bubble sort
import random

def main():
    lst = []

    for i in range(30):
        lst.append(random.randint(100, 999))

    print(lst)
    bubble_sort(lst)
    print(lst)


def bubble_sort(input_list):
    """
    i is index being checked (left to right) to swap with index to the right of i only when i is greater
    if i is the greatest in the list, it will bubble to all the way to the right
    if i is not the greatest, it will only bubble to the right stopping until the index to the right is greater...loop continues however i will reflect the next index to the right as ussual(meaning different value to bubble). Previous i may bubble higher on future loops
    """
    did_swap = True
    sort_cnt = 1#doesn't need to be zero since the last item isn't ever swapped to the right
    while did_swap:#better than for loop for early detection of pre-sorted list
        did_swap = False

        for i in range(len(input_list) - sort_cnt):
            if input_list[i] > input_list[i + 1]:#swaps index being checked with index to the right if greater
                input_list[i], input_list[i + 1] = input_list[i + 1], input_list[i]
                did_swap = True#keeps outter loop sorting until nothing bubbles to the right

        sort_cnt += 1#shortens the needed range to search for greater value since greatest values have already been sorted to the right


main()
```
